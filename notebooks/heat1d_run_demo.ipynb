```python
# run_demo.py

import sys
import feelpp.core as fppc
import feelpp.toolboxes.core as tb
from feelpp.scimba.heat1d_solver import FEM_PINN_Solver

sys.argv = ["feelpp_fem"]
env = fppc.Environment(
    sys.argv, 
    opts=tb.toolboxes_options("coefficient-form-pdes", "cfpdes"), 
    config=fppc.localRepository("fem-repo")
)

solver = FEM_PINN_Solver(xm=0.4, ω=0.8, max_iter=5, tol=1e-4, alpha=0.5)
t, u = solver.solve()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
alpha = 0.5
times = [0.0, 0.006, 0.014, 0.020, 0.028]
x = np.linspace(0, 1, 500)

plt.figure(figsize=(8, 4))
for t in times:
    u_exact = np.exp(-alpha * np.pi**2 * t) * np.sin(np.pi * x)
    plt.plot(x, u_exact, label=f"t={t:.3f}")

plt.xlabel("x")
plt.ylabel("u_exact")
plt.title("Exact Solution at Selected Times")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
